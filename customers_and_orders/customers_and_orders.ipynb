**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

customers = []

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # skip header row

    for row in csvreader:
        if not row or len(row) < 5:
            continue  # skip empty/malformed rows

        customer_id, first_name, last_name, city, state = [c.strip() for c in row[:5]]

        if not customer_id:
            continue  # skip rows with no id

        customers.append({
            "CustomerId": customer_id,
            "FirstName": first_name,
            "LastName": last_name,
            "City": city,
            "State": state
        })

num_customers = len(customers)
print(f"Number of customers: {num_customers}")
```

**Question 2:** In how many different states do the customers live in?

```python
states = set(c["State"].upper() for c in customers if c["State"])
num_states = len(states)
print(f"Number of different states: {num_states}")
```

**Question 3** What is the state with most customers?

```python
from collections import Counter

state_counts = Counter(c["State"].upper() for c in customers if c["State"])
most_common_state, most_common_count = state_counts.most_common(1)[0]
print(f"State with most customers: {most_common_state} ({most_common_count} customers)")
```

**Question 4** What is the state with the least customers?

```python
min_count = min(state_counts.values())
states_with_min = sorted(state for state, count in state_counts.items() if count == min_count)

print(f"Minimum customer count: {min_count}")
print(f"State(s) with least customers: {', '.join(states_with_min)}")
```

**Question 5:** What is the most common last name?

```python
last_name_counts = Counter(c["LastName"].strip().upper() for c in customers if c["LastName"])
most_common_last_name, most_common_last_name_count = last_name_counts.most_common(1)[0]
print(f"Most common last name: {most_common_last_name} ({most_common_last_name_count} occurrences)")
```

```python
from os.path import exists
import csv
from collections import defaultdict
from dateutil import parser as dateparser

orders_datafile = "./sample_data/orders.csv"
if not exists(orders_datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")


def parse_date(date_str):
    """Parse a date string in any common format; return None if invalid/empty."""
    date_str = date_str.strip()
    if not date_str:
        return None
    try:
        return dateparser.parse(date_str)
    except (ValueError, TypeError):
        return None


def parse_amount(value):
    """Parse a currency string (handles '$', ',', whitespace) into a float."""
    try:
        return float(value.strip().replace("$", "").replace(",", ""))
    except (ValueError, AttributeError):
        return None


# order_items[order_id]  -> number of product rows for that order
# order_info[order_id]   -> {"CustomerID", "Date", "OrderTotal"} (captured once per order)
order_items = defaultdict(int)
order_info = {}

with open(orders_datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # skip header

    for row in csvreader:
        if not row or len(row) < 6:
            continue

        customer_id, order_id, date_str, order_total_str, product_name, price_str = [c.strip() for c in row[:6]]

        if not order_id or not customer_id:
            continue

        order_items[order_id] += 1

        # Capture order-level fields only once per order to avoid re-summing OrderTotal per product row
        if order_id not in order_info:
            order_info[order_id] = {
                "CustomerID": customer_id,
                "Date": parse_date(date_str),
                "OrderTotal": parse_amount(order_total_str) or 0.0,
            }

print(f"Parsed {len(order_info)} unique orders from {sum(order_items.values())} product rows.")
```

```python
num_unique_orders = len(order_items)
print(f"Q1 - Unique orders: {num_unique_orders}")
```

```python
total_item_rows = sum(order_items.values())
avg_items_per_order = round(total_item_rows / num_unique_orders, 2)
print(f"Q2 - Average items per order: {avg_items_per_order}")
```

```python
max_items_per_order = max(order_items.values())
print(f"Q3 - Highest number of items in one order: {max_items_per_order}")
```

```python
october_2021_orders = sum(
    1 for info in order_info.values()
    if info["Date"] and info["Date"].year == 2021 and info["Date"].month == 10
)
print(f"Q4 - Orders placed in October 2021: {october_2021_orders}")
```

```python
sales_by_month = defaultdict(float)
orders_by_month = defaultdict(int)

for info in order_info.values():
    if info["Date"]:
        month_name = info["Date"].strftime("%B")
        sales_by_month[month_name] += info["OrderTotal"]
        orders_by_month[month_name] += 1

best_month, best_month_sales = max(sales_by_month.items(), key=lambda x: x[1])
print(f"Q6 - Best month for sales: {best_month} (${best_month_sales:.2f})")

print("\nAll months ranked by total sales:")
for month, total in sorted(sales_by_month.items(), key=lambda x: x[1], reverse=True):
    print(f"  {month}: ${total:.2f} ({orders_by_month[month]} orders)")
```
